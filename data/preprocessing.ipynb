```python
import pandas as pd
import numpy as np
```

```python
metadata = pd.read_csv('/home/atc009/biocurate/data/pfs_months.txt', sep='\t')
metadata
```

```python
survival = pd.read_csv('/home/atc009/biocurate/data/OS_time_adjusted.txt', sep='\t')
survival_treatment = pd.merge(survival, metadata[['PATIENT_ID', 'DRUG']], on='PATIENT_ID')
survival_treatment
```

```python
patient2ind = pd.read_csv('/home/atc009/biocurate/data/patient2ind_msk.txt', sep='\t')
patient2ind = patient2ind.rename(columns={'sample':'SAMPLE_ID'})
patient2ind.to_csv('/home/atc009/biocurate/data/patient2ind.txt', sep='\t', index=None)

final_metadata = survival_treatment.merge(patient2ind, on='SAMPLE_ID')
final_metadata
```

```python
final_metadata.to_csv('/home/atc009/biocurate/data/patient_metadata.txt', sep='\t', index=None)
```
